```python
a = np.array([2, 5, 6])
b = np.array([3, 0, 2])

print('a . b  (dot)   :', np.dot(a, b))

print('a x b  (cross) :', np.cross(a, b))
```

```python
print('L2 norm  ||a||_2 :', la.norm(a))
print('L1 norm  ||a||_1 :', la.norm(a, 1))
print('Linf norm        :', la.norm(a, np.inf))

def cosine(u, v):
    return np.dot(u, v) / (la.norm(u) * la.norm(v))

print('cosine(a, b)     :', round(cosine(a, b), 3))
```

```python
pairs = [
    (np.array([1, 0, 1]), np.array([1, 0, 1])),
    (np.array([1, 2, 3]), np.array([3, 2, 1])),
    (np.array([2, 0, 0]), np.array([0, 5, 0])),
]

similarities = []

for i, (u, v) in enumerate(pairs, start=1):

    L1 = np.linalg.norm(u, ord=1)
    L2 = np.linalg.norm(u, ord=2)

    cosine = np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))

    similarities.append(cosine)

    print(f"Pair {i}:")
    print("L1 norm of u:", L1)
    print("L2 norm of u:", L2)
    print("Cosine similarity:", cosine)
    print()

most_similar = np.argmax(similarities) + 1
print("Most similar pair =", most_similar)
```

```python
A = np.array([[2., 1.],
              [1., 3.]])
B = np.array([[1., 0.],
              [4., 2.]])

print('A @ B (matrix multiply):\n', A @ B)
print('A.T   (transpose):\n', A.T)
print('inverse(A):\n', la.inv(A))
print('trace(A) = sum of diagonal:', np.trace(A))
```

```python
I = np.eye(2)
print('Identity I:\n', I)
print('A @ inv(A) == I ?', np.allclose(A @ la.inv(A), I))

print('A symmetric?', np.allclose(A, A.T))

theta = np.radians(30)
Q = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
print('Q orthogonal?', np.allclose(Q.T @ Q, I))
```

```python
import numpy as np

M = np.array([[4., 2.],
              [2., 3.]])
P = np.array([[0., -1.],
              [1.,  0.]])

inv_M = np.linalg.inv(M)
print("Inverse of M:\n", inv_M)

print("Trace of M:", np.trace(M))

result = M @ inv_M
print("M @ inv(M):\n", result)
print("M @ inv(M) == I ?", np.allclose(result, np.eye(2)))


print("Is M symmetric?", np.allclose(M, M.T))

print("Is P orthogonal?", np.allclose(P.T @ P, np.eye(2)))
```

```python
square = np.array([[0, 1, 1, 0],
                   [0, 0, 1, 1]], dtype=float)

S = np.array([[1.5, 0.0],
              [0.0, 0.5]])

t = np.radians(30)
R = np.array([[np.cos(t), -np.sin(t)],
              [np.sin(t),  np.cos(t)]])

scaled  = S @ square
rotated = R @ square
print('Rotated corners:\n', rotated)
```

```python
def close_loop(pts):
    return np.hstack([pts, pts[:, :1]])

fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(*close_loop(square),  marker='o', label='original')
ax.plot(*close_loop(scaled),  marker='o', label='scaled')
ax.plot(*close_loop(rotated), marker='o', label='rotated 30 deg')
ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
ax.set_aspect('equal'); ax.legend(); ax.set_title('Linear transformations')
plt.show()
```

```python
tri = np.array([[0, 1, 0.5],
                [0, 0, 1.0]])
S_tri = np.array([[2.0, 0.0],
                  [0.0, 0.5]])
theta_tri = np.radians(45)
R_tri = np.array([[np.cos(theta_tri), -np.sin(theta_tri)],
                  [np.sin(theta_tri),  np.cos(theta_tri)]])
scaled_tri  = S_tri @ tri
rotated_tri = R_tri @ tri

fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(*close_loop(tri),       marker='o', label='original triangle')
ax.plot(*close_loop(scaled_tri),  marker='o', label='scaled triangle')
ax.plot(*close_loop(rotated_tri), marker='o', label='rotated 45 deg triangle')
ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
ax.set_aspect('equal'); ax.legend(); ax.set_title('Triangle Transformations')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
A = np.array([[5., 0.],
              [0., 6.]])

vals, vecs = la.eig(A)
print('Eigenvalues  (lambda):', vals)
print('Eigenvectors (columns):\n', vecs)
```

```python
for i in range(len(vals)):
    v = vecs[:, i]
    lhs = A @ v
    rhs = vals[i] * v
    print(f'lambda={vals[i]:.1f}  A v == lambda v ?', np.allclose(lhs, rhs))
```

```python
C = np.array([[4., 1.],
              [2., 3.]])

vals_C, vecs_C = la.eig(C)

print('Eigenvalues of C (lambda):', vals_C)
print('Eigenvectors of C (columns):\n', vecs_C)

v1 = vecs_C[:, 0]
lambda1 = vals_C[0]
lhs_C = C @ v1
rhs_C = lambda1 * v1
print(f'\nFor first eigenvector (lambda={lambda1:.3f}): C v == lambda v ?', np.allclose(lhs_C, rhs_C))
```
